```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
from sklearn.decomposition import PCA
```

```python
def remove_all_zero_col(data):
    data = data.copy()
    for col in data.columns:
        if (data[col] == 0).all():
            data.drop(col, axis=1, inplace=True)
    return data
```

```python
data = pd.read_csv("../source/smiles.csv")
Yield = pd.DataFrame(data['Yield'], columns=['Yield'])
PC_ID = pd.DataFrame(data['PC_ID'], columns=['PC_ID'])
Alc_ID = pd.DataFrame(data['Alc_ID'], columns=['Alc_ID'])
smiles_PC = pd.DataFrame(data['PC_SMILES'], columns=['PC_SMILES'])
smiles_Alc = pd.DataFrame(data['Alc_SMILES'], columns=['Alc_SMILES'])

print(Yield)
print(PC_ID)
print(Alc_ID)
```

```python
#PC
mf_p = []
for i in range(0,196):
    mol_p = Chem.MolFromSmiles(data["PC_SMILES"][i])
    fp_p = AllChem.GetMorganFingerprintAsBitVect(mol_p, 2, 2048)
    mf_p.append(np.array(fp_p, int))

df_p = pd.DataFrame(mf_p)
df_p = df_p.add_prefix('PC_MF_')
print(df_p.shape)
df_2_p = remove_all_zero_col(df_p)
print(df_2_p.shape)
df_2_p.head()
```

```python
#Alc
mf_a = []
for i in range(0,196):
    mol_a = Chem.MolFromSmiles(data["Alc_SMILES"][i])
    fp_a = AllChem.GetMorganFingerprintAsBitVect(mol_a, 2, 2048)
    mf_a.append(np.array(fp_a, int))

df_a = pd.DataFrame(mf_a)
df_a = df_a.add_prefix('Alc_MF_')
print(df_a.shape)
df_2_a = remove_all_zero_col(df_a)
print(df_2_a.shape)
df_2_a.head()
```

```python
MF = pd.concat([PC_ID, smiles_PC, df_2_p, Alc_ID, smiles_Alc, df_2_a, Yield], axis=1, join='inner')
MF.to_csv('../data/data_MF.csv', index = False)
MF
```

```python
#PC, PCA
X_pc = np.array(df_2_p)
pca = PCA(n_components=7)
X_pca_pc = pca.fit(X_pc).transform(X_pc)
print("Before: {}".format(str(X_pc.shape)))
print("After: {}".format(str(X_pca_pc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_3_p = pd.DataFrame(X_pca_pc)
df_3_p = df_3_p.add_prefix('MF_PC_')
df_3_p.head()
```

```python
#Alc, PCA
X_alc = np.array(df_2_a)
pca = PCA(n_components=7)
X_pca_alc = pca.fit(X_alc).transform(X_alc)
print("Before: {}".format(str(X_alc.shape)))
print("After: {}".format(str(X_pca_alc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_3_a = pd.DataFrame(X_pca_alc)
df_3_a = df_3_a.add_prefix('MF_Alc_')
df_3_a.head()
```

```python
MF_pca = pd.concat([PC_ID, smiles_PC, df_3_p, Alc_ID, smiles_Alc, df_3_a, Yield], axis=1, join='inner')
MF_pca.to_csv('../data/data_MF_pca.csv', index = False)
MF_pca
```
