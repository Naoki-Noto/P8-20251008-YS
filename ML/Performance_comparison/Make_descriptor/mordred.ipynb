```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
from mordred import Calculator, descriptors
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
def remove_all_zero_col(data):
    data = data.copy()
    for col in data.columns:
        if (data[col] == 0).all():
            data.drop(col, axis=1, inplace=True)
    return data
```

```python
data = pd.read_csv("../source/smiles.csv")
Yield = pd.DataFrame(data['Yield'], columns=['Yield'])
PC_ID = pd.DataFrame(data['PC_ID'], columns=['PC_ID'])
Alc_ID = pd.DataFrame(data['Alc_ID'], columns=['Alc_ID'])
smiles_PC = pd.DataFrame(data['PC_SMILES'], columns=['PC_SMILES'])
smiles_Alc = pd.DataFrame(data['Alc_SMILES'], columns=['Alc_SMILES'])

print(Yield)
print(PC_ID)
print(Alc_ID)
```

```python
#PC
data['Mol'] = data["PC_SMILES"].apply(Chem.MolFromSmiles)

calc_mordred_p = Calculator(descriptors, ignore_3D=True)
desc_mordred_p = calc_mordred_p.pandas(data['Mol'])

mordred_p = desc_mordred_p.astype(str)
masks_p = mordred_p.apply(lambda d: d.str.contains('[a-zA-Z]' ,na=False))
mordred_p = mordred_p[~masks_p]
mordred_p = mordred_p.astype(float)
df_mordred_p = mordred_p.dropna(how="any", axis="columns")

print(df_mordred_p.shape)
df_mordred_p_2 = remove_all_zero_col(df_mordred_p)
df_mordred_p_2 = df_mordred_p_2.add_prefix('PC_')
print(df_mordred_p_2.shape)
df_mordred_p_2.head()
```

```python
#Alc
data['Mol'] = data["Alc_SMILES"].apply(Chem.MolFromSmiles)

calc_mordred_a = Calculator(descriptors, ignore_3D=True)
desc_mordred_a = calc_mordred_a.pandas(data['Mol'])

mordred_a = desc_mordred_a.astype(str)
masks_a = mordred_a.apply(lambda d: d.str.contains('[a-zA-Z]' ,na=False))
mordred_a = mordred_a[~masks_a]
mordred_a = mordred_a.astype(float)
df_mordred_a = mordred_a.dropna(how="any", axis="columns")

print(df_mordred_a.shape)
df_mordred_a_2 = remove_all_zero_col(df_mordred_a)
df_mordred_a_2 = df_mordred_a_2.add_prefix('Alc_')
print(df_mordred_a_2.shape)
df_mordred_a_2.head()
```

```python
mordred = pd.concat([PC_ID, smiles_PC, df_mordred_p_2, Alc_ID, smiles_Alc, df_mordred_a_2, Yield], axis=1, join='inner')
mordred.to_csv('../data/data_mordred.csv', index = False)
mordred
```

```python
#PC, PCA
X_pc = np.array(df_mordred_p_2)
sc = StandardScaler()
X_sc_pc = sc.fit(X_pc).transform(X_pc)
pca = PCA(n_components=5)
X_pca_pc = pca.fit(X_sc_pc).transform(X_sc_pc)
print("Before: {}".format(str(X_sc_pc.shape)))
print("After: {}".format(str(X_pca_pc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_mordred_p_3 = pd.DataFrame(X_pca_pc)
df_mordred_p_3 = df_mordred_p_3.add_prefix('mordred_PC_')
df_mordred_p_3.head()
```

```python
#Alc, PCA
X_alc = np.array(df_mordred_a_2)
sc = StandardScaler()
X_sc_alc = sc.fit(X_alc).transform(X_alc)
pca = PCA(n_components=5)
X_pca_alc = pca.fit(X_sc_alc).transform(X_sc_alc)
print("Before: {}".format(str(X_sc_alc.shape)))
print("After: {}".format(str(X_pca_alc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_mordred_a_3 = pd.DataFrame(X_pca_alc)
df_mordred_a_3 = df_mordred_a_3.add_prefix('mordred_Alc_')
df_mordred_a_3.head()
```

```python
mordred_pca = pd.concat([PC_ID, smiles_PC, df_mordred_p_3, Alc_ID, smiles_Alc, df_mordred_a_3, Yield], axis=1, join='inner')
mordred_pca.to_csv('../data/data_mordred_pca.csv', index = False)
mordred_pca
```
