```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import PandasTools
from rdkit.Chem import Descriptors
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import warnings
```

```python
def remove_all_zero_col(data):
    data = data.copy()
    for col in data.columns:
        if (data[col] == 0).all():
            data.drop(col, axis=1, inplace=True)
    return data
```

```python
warnings.filterwarnings('ignore', category=pd.errors.PerformanceWarning)
```

```python
data = pd.read_csv("../source/smiles.csv")
Yield = pd.DataFrame(data['Yield'], columns=['Yield'])
PC_ID = pd.DataFrame(data['PC_ID'], columns=['PC_ID'])
Alc_ID = pd.DataFrame(data['Alc_ID'], columns=['Alc_ID'])
df_PC = pd.DataFrame(data['PC_SMILES'], columns=['PC_SMILES'])
df_Alc = pd.DataFrame(data['Alc_SMILES'], columns=['Alc_SMILES'])
smiles_PC = pd.DataFrame(data['PC_SMILES'], columns=['PC_SMILES'])
smiles_Alc = pd.DataFrame(data['Alc_SMILES'], columns=['Alc_SMILES'])

print(Yield)
print(PC_ID)
print(Alc_ID)
```

```python
#PC
PandasTools.AddMoleculeColumnToFrame(df_PC,'PC_SMILES')
for i,j in Descriptors.descList:
    df_PC[i] = df_PC.ROMol.map(j)
df_PC = df_PC.add_prefix('PC_')

df_p = df_PC.drop(columns=['PC_PC_SMILES', 'PC_ROMol'])
print(df_p.shape)
df_2_p = remove_all_zero_col(df_p)
df_2_p = df_2_p.dropna(how='any', axis=1)
print(df_2_p.shape)
df_2_p.head()
```

```python
#Alc
PandasTools.AddMoleculeColumnToFrame(df_Alc,'Alc_SMILES')
for i,j in Descriptors.descList:
    df_Alc[i] = df_Alc.ROMol.map(j)
df_Alc = df_Alc.add_prefix('Alc_')

df_a = df_Alc.drop(columns=['Alc_Alc_SMILES', 'Alc_ROMol'])
print(df_a.shape)
df_2_a = remove_all_zero_col(df_a)
df_2_a = df_2_a.dropna(how='any', axis=1)
print(df_2_a.shape)
df_2_a.head()
```

```python
RDKit = pd.concat([PC_ID, smiles_PC, df_2_p, Alc_ID, smiles_Alc, df_2_a, Yield], axis=1, join='inner')
RDKit.to_csv('../data/data_RDKit.csv', index = False)
RDKit
```

```python
#PC, PCA
X_pc = np.array(df_2_p)
sc = StandardScaler()
X_sc_pc = sc.fit(X_pc).transform(X_pc)
pca = PCA(n_components=7)
X_pca_pc = pca.fit(X_sc_pc).transform(X_sc_pc)
print("Before: {}".format(str(X_sc_pc.shape)))
print("After: {}".format(str(X_pca_pc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_3_p = pd.DataFrame(X_pca_pc)
df_3_p = df_3_p.add_prefix('RDKit_PC_')
df_3_p.head()
```

```python
#Alc, PCA
X_alc = np.array(df_2_a)
sc = StandardScaler()
X_sc_alc = sc.fit(X_alc).transform(X_alc)
pca = PCA(n_components=5)
X_pca_alc = pca.fit(X_sc_alc).transform(X_sc_alc)
print("Before: {}".format(str(X_sc_alc.shape)))
print("After: {}".format(str(X_pca_alc.shape)))
print('sum of explained variance ratio: {0}'.format(sum(pca.explained_variance_ratio_)))
df_3_a = pd.DataFrame(X_pca_alc)
df_3_a = df_3_a.add_prefix('RDKit_Alc_')
df_3_a.head()
```

```python
RDKit_pca = pd.concat([PC_ID, smiles_PC, df_3_p, Alc_ID, smiles_Alc, df_3_a, Yield], axis=1, join='inner')
RDKit_pca.to_csv('../data/data_RDKit_pca.csv', index = False)
RDKit_pca
```
