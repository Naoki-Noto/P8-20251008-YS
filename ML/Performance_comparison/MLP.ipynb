```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import VarianceThreshold
from sklearn import metrics
```

```python
data = pd.read_csv('data/data_DFT.csv')
y = pd.DataFrame(data['Yield'], columns=['Yield'])
X = data.drop(columns=['PC_ID', 'Alc_ID', 'Yield', 'PC_SMILES', 'Alc_SMILES'])
print(X.head())
print('=======================================================================')
pipe = Pipeline([("var", VarianceThreshold(threshold=0.0)),
                 ("scaler", StandardScaler()),
                 ("mlp", MLPRegressor(random_state=0, max_iter=1500, learning_rate_init=0.03)),
                ])

param = {'mlp__hidden_layer_sizes': [(128, 128), (256, 256), (512, 512)],
         'mlp__alpha': [0, 1, 2, 5],
        }

r2_list = []
rmse_list = []
for i in range(10):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=i)
    reg = GridSearchCV(estimator=pipe, param_grid=param, cv=5, n_jobs=12)
    reg.fit(X_train, y_train['Yield'])
    best = reg.best_estimator_
    y_pred1 = best.predict(X_train)
    y_pred2 = best.predict(X_test)
    r2 = metrics.r2_score(y_test, y_pred2)
    rmse = metrics.root_mean_squared_error(y_test, y_pred2)
    print(f'Run{i} R2 (test): {r2:.2f}, RMSE (test): {rmse:.2f}')
    r2_list.append(r2)
    rmse_list.append(rmse)
print('==========(Result)==========')
print(f'Mean R2: {np.mean(r2_list):.3f}')
print(f'SD R2: {np.std(r2_list, ddof=1):.3f}')
print(f'Mean RMSE: {np.mean(rmse_list):.3f}')
print(f'SD RMSE: {np.std(rmse_list, ddof=1):.3f}')
```
