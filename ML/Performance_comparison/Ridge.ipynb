```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import RidgeCV
from sklearn import metrics
```

```python
data = pd.read_csv('data/data_DFT.csv')
y = pd.DataFrame(data['Yield'], columns=['Yield'])
X = data.drop(columns=['PC_ID', 'Alc_ID', 'Yield', 'PC_SMILES', 'Alc_SMILES'])
print(X.head())
print('=======================================================================')

r2_score = []
rmse_score = []
for i in range(10):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=i)
    a_X_train = (X_train - X_train.mean()) / X_train.std()
    a_X_test = (X_test - X_train.mean()) / X_train.std()
    a_X_train = a_X_train.dropna(how='any', axis=1)
    a_X_test = a_X_test[a_X_train.columns]
    reg = RidgeCV(alphas=np.linspace(0.1, 30, num=150), cv=5)
    reg.fit(a_X_train, y_train['Yield'])
    y_pred1 = reg.predict(a_X_train)
    y_pred2 = reg.predict(a_X_test)
    r2 = metrics.r2_score(y_test, y_pred2)
    rmse = metrics.root_mean_squared_error(y_test, y_pred2)
    print(f'Run{i} R2 (test): {r2:.2f}, RMSE (test): {rmse:.2f}')
    r2_score.append(r2)
    rmse_score.append(rmse)
print('==========(Result)==========')
print(f'Mean R2: {np.mean(r2_score):.3f}')
print(f'SD R2: {np.std(r2_score):.3f}')
print(f'Mean RMSE: {np.mean(rmse_score):.3f}')
print(f'SD RMSE: {np.std(rmse_score):.3f}')
```
