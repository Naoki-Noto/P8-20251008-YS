```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import RidgeCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import VarianceThreshold
from sklearn import metrics
```

```python
data = pd.read_csv('data/data_DFT.csv')
y = pd.DataFrame(data['Yield'], columns=['Yield'])
X = data.drop(columns=['PC_ID', 'Alc_ID', 'Yield', 'PC_SMILES', 'Alc_SMILES'])
print(X.head())
print('=======================================================================')

pipe = Pipeline([("var", VarianceThreshold(threshold=0.0)),
                 ("scaler", StandardScaler()),
                 ("ridge", RidgeCV(alphas=np.linspace(0.1, 30, num=150), cv=5)),
                ])

r2_list = []
rmse_list = []
for i in range(10):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=i)
    pipe.fit(X_train, y_train['Yield'])
    y_pred_train = pipe.predict(X_train)
    y_pred_test = pipe.predict(X_test)
    r2 = metrics.r2_score(y_test, y_pred_test)
    rmse = metrics.root_mean_squared_error(y_test, y_pred_test)
    print(f'Run{i} R2 (test): {r2:.2f}, RMSE (test): {rmse:.2f}')
    r2_list.append(r2)
    rmse_list.append(rmse)
print('==========(Result)==========')
print(f'Mean R2: {np.mean(r2_list):.3f}')
print(f'SD R2: {np.std(r2_list):.3f}')
print(f'Mean RMSE: {np.mean(rmse_list):.3f}')
print(f'SD RMSE: {np.std(rmse_list):.3f}')
```
